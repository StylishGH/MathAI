```python
import re

def padronizar_banca(nome_raw):
    nome = str(nome_raw).strip()
    if not nome or nome == 'nan':
        return 'Olimpíada / Internacional'

    # 1. Se tiver "TST" ou "Selektion" ou "Selection", é uma SELETIVA NACIONAL do país:
    if re.search(r'TST|Selektion|Selection', nome, re.I):
        if re.search(r'India', nome, re.I):
            return 'Índia (TST)'
        if re.search(r'HK|Hong Kong', nome, re.I):
            return 'Hong Kong (TST)'
        return f'{nome} (Seletiva)'

    # 2. Se for a IMO Internacional DE VERDADE:
    if re.search(r'\bIMO\b|International Mathematical Olympiad', nome, re.I):
        return 'IMO (Oficial)'

    # 3. Mapeamento das Grandes Olimpíadas
    if re.search(r'Harvard-MIT|HMMT', nome, re.I):
        return 'HMMT (Harvard-MIT)'
    if re.search(r'Brazilian|OBM|Nível [123]', nome, re.I):
        return 'OBM (Brasil)'
    if re.search(r'Comunidade dos Países de Língua Portuguesa|OMCPLP', nome, re.I):
        return 'OMCPLP (Países Lusófonos)'
    if re.search(r'Iberoamerican|Ibero-americana', nome, re.I):
        return 'OIM (Ibero-Americana)'
    if re.search(r'Junior Balkan|JBMO', nome, re.I):
        return 'JBMO (Balcãs Júnior)'
    if re.search(r'Balkan|BMO', nome, re.I):
        return 'BMO (Balcãs)'
    if re.search(r'Asia Pacific|APMO', nome, re.I):
        return 'APMO (Ásia-Pacífico)'
    if re.search(r'European Girls|EGMO', nome, re.I):
        return 'EGMO (Feminina Europeia)'
    if re.search(r'Baltic Way', nome, re.I):
        return 'Baltic Way'
    if re.search(r'Françaises|Française|POFM', nome, re.I):
        return 'França (POFM)'
    if re.search(r'Iranian|Iran', nome, re.I):
        return 'Irã MO'
    if re.search(r'Canadian|CMO', nome, re.I):
        return 'Canadá (CMO)'
    if re.search(r'Russian|All-Russian', nome, re.I):
        return 'Rússia MO'
    if re.search(r'Japan|JMO', nome, re.I):
        return 'Japão (JMO)'
    if re.search(r'China|Chinese', nome, re.I):
        return 'China MO'
    if re.search(r'Berkeley Math Circle', nome, re.I):
        return 'Berkeley Math Circle'
    if re.search(r'Romanian|RMM', nome, re.I):
        return 'Romênia (RMM)'
    if re.search(r'Putnam', nome, re.I):
        return 'Putnam (EUA/Universitária)'

    # 4. Para qualquer outra banca: remove anos e edições
    banca_limpa = re.sub(r'\b(19\d{2}|20\d{2})\b', '', nome)
    banca_limpa = re.sub(r'\b\d+(st|nd|rd|th)\b', '', banca_limpa, flags=re.I)
    banca_limpa = re.sub(r'\b(Nível|Edition|Edição)\s*\d+\b', '', banca_limpa, flags=re.I)
    banca_limpa = banca_limpa.strip(' -_:,()')

    return banca_limpa if banca_limpa else 'Olimpíada'

# Aplica no DataFrame inteiro de uma vez só:
df['banca'] = df['banca'].apply(padronizar_banca)
```

## 🏷️ Classificar Questões no DataFrame (Objetiva vs Discursiva)

Classifica as questões **no pandas** antes de inserir no banco.
O campo `tipo` é calculado automaticamente — ou você pode sobrescrever manualmente no df.

> **Regra:** tem alternativas `(A)(B)...` ou gabarito é letra A–E → `objetiva` | caso contrário → `discursiva`

```python
import re

def classificar_tipo(enunciado: str, gabarito: str) -> str:
    tem_alternativas = bool(re.search(r'(?m)^\(([A-E])\)\s*(.+)$', enunciado or ""))
    gabarito_e_letra = str(gabarito or "").strip().upper() in ("A", "B", "C", "D", "E")
    return "objetiva" if (tem_alternativas or gabarito_e_letra) else "discursiva"

df['tipo'] = df.apply(lambda r: classificar_tipo(r['enunciado'], r['gabarito']), axis=1)

print(df['tipo'].value_counts())
display(df[['id', 'banca', 'ano', 'gabarito', 'tipo']].head(20))
```

## 📥 Extração e Importação de Questões do ENEM (via biblioteca `enem`)

Esta seção utiliza o pacote **Enem Extractor** (`enem`) para:
1. Ler o PDF oficial de uma prova do ENEM (Dia 2 - Matemática e Ciências da Natureza).
2. Extrair o texto das questões, alternativas (A, B, C, D, E) e o gabarito oficial em JSON.
3. Recortar e salvar automaticamente as **figuras e gráficos** das questões.
4. Inserir as questões de Matemática (136 a 180) diretamente no banco de dados do **MathAI** (`questoes` no Turso/SQLite).


```python
# Instala a biblioteca enem se ainda não estiver instalada
#!pip install enem
import os
import sys
import json
import shutil
import subprocess
from pathlib import Path

# Garante acesso à raiz do projeto
BASE_DIR = Path.cwd().parent
sys.path.append(str(BASE_DIR))

from src.database.db import pegar_conexao
print("✅ Ambiente configurado para importação do ENEM!")
```

```python
def extrair_e_importar_enem(
    caminho_pdf_prova: str,
    caminho_pdf_gabarito: str | None = None,
    ano: int = 2023,
    pasta_saida: str = "../data/enem_extraido",
    questoes_apenas_matematica: bool = True
):
    """
    Executa a extração do ENEM via CLI do 'enem', processa o JSON gerado,
    copia as imagens para data/uploads/ e insere no banco de dados do MathAI.
    """
    pasta_saida_path = Path(pasta_saida)
    pasta_saida_path.mkdir(parents=True, exist_ok=True)
    pasta_uploads = BASE_DIR / "data" / "uploads"
    pasta_uploads.mkdir(parents=True, exist_ok=True)

    cmd = [sys.executable, "-m", "enem", str(caminho_pdf_prova), "-o", str(pasta_saida_path)]
    if caminho_pdf_gabarito:
        cmd.extend(["-g", str(caminho_pdf_gabarito)])

    print(f"🚀 Executando Enem Extractor para o ENEM {ano}...")
    resultado = subprocess.run(cmd, capture_output=True, text=True)
    if resultado.returncode != 0:
        print(f"❌ Erro na extração do ENEM:\n{resultado.stderr}")
        return 0

    print("✅ Extração do PDF concluída com sucesso! Processando questões...")

    # Localiza o arquivo JSON gerado na pasta de saída
    arquivos_json = list(pasta_saida_path.glob("*.json"))
    if not arquivos_json:
        print("⚠️ Nenhum arquivo JSON encontrado na pasta de saída.")
        return 0

    with open(arquivos_json[0], "r", encoding="utf-8") as f:
        dados_enem = json.load(f)

    con = pegar_conexao()
    cur = con.cursor()
    total_inseridas = 0

    questoes_lista = dados_enem if isinstance(dados_enem, list) else dados_enem.get("questions", [])

    for q in questoes_lista:
        num = q.get("number") or q.get("index") or 0
        # No 2º dia do ENEM (a partir de 2009), Matemática vai da questão 136 até a 180
        if questoes_apenas_matematica and not (136 <= int(num) <= 180):
            continue

        enunciado = q.get("statement", "") or q.get("enunciado", "") or q.get("text", "")
        alternativas = q.get("alternatives", {}) or q.get("options", {})
        
        texto_completo = enunciado.strip()
        if alternativas:
            texto_completo += "\n\n"
            for letra in ["A", "B", "C", "D", "E"]:
                if letra in alternativas:
                    texto_completo += f"({letra}) {alternativas[letra]}\n"

        gabarito = str(q.get("answer") or q.get("gabarito") or "").strip().upper()

        figura_relativa = None
        imagem_origem = q.get("image") or q.get("figure")
        if imagem_origem:
            caminho_img = pasta_saida_path / imagem_origem
            if caminho_img.exists():
                nome_destino = f"enem_{ano}_q{num}_{caminho_img.name}"
                destino = pasta_uploads / nome_destino
                shutil.copy2(caminho_img, destino)
                figura_relativa = f"data/uploads/{nome_destino}"

        try:
            cur.execute("""
                INSERT INTO questoes (
                    materia, topico, enunciado, gabarito, dificuldade,
                    banca, ano, figura_path, tipo
                ) VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
            """, (
                "Matemática", "ENEM", texto_completo, gabarito, 3,
                "ENEM", ano, figura_relativa, "objetiva"
            ))
            total_inseridas += 1
        except Exception as e:
            print(f"Erro ao inserir questão {num}: {e}")

    con.commit()
    print(f"🎉 ENEM {ano}: {total_inseridas} questões de Matemática importadas com sucesso!\n")
    return total_inseridas


def importar_enem_por_ano(
    ano: int,
    pasta_provas: str = "../data/provas_enem",
    pasta_saida: str = "../data/enem_extraido"
):
    """
    Localiza automaticamente os PDFs da prova e gabarito do ano especificado
    e executa a extração e importação.
    Exemplo de uso em loop:
        for ano in range(2009, 2025):
            importar_enem_por_ano(ano)
    """
    pasta = Path(pasta_provas)
    if not pasta.exists():
        pasta.mkdir(parents=True, exist_ok=True)
        print(f"📁 Pasta criada: {pasta.resolve()}. Coloque os PDFs lá dentro!")
        return 0

    # Procura arquivos PDF do ano especificado
    arquivos_ano = list(pasta.glob(f"*{ano}*.pdf"))
    if not arquivos_ano:
        print(f"⚠️ [ENEM {ano}] Nenhum PDF encontrado contendo '{ano}' em {pasta}. Pulando...")
        return 0

    # Identifica o gabarito e a prova
    gabarito_pdf = None
    prova_pdf = None

    for f in arquivos_ano:
        nome_lower = f.name.lower()
        if "gabarito" in nome_lower or "gab" in nome_lower or "key" in nome_lower:
            gabarito_pdf = str(f)
        elif "dia2" in nome_lower or "matematica" in nome_lower or "mat" in nome_lower or "caderno" in nome_lower or "prova" in nome_lower:
            prova_pdf = str(f)

    # Se não achou prova específica com dia2/mat, pega o primeiro que não for gabarito
    if not prova_pdf:
        candidatos = [str(f) for f in arquivos_ano if str(f) != gabarito_pdf]
        if candidatos:
            prova_pdf = candidatos[0]

    if not prova_pdf:
        print(f"⚠️ [ENEM {ano}] Não foi possível identificar o PDF da prova. Pulando...")
        return 0

    print(f"📄 [ENEM {ano}] Prova: {Path(prova_pdf).name} | Gabarito: {Path(gabarito_pdf).name if gabarito_pdf else 'Nenhum'}")
    return extrair_e_importar_enem(
        caminho_pdf_prova=prova_pdf,
        caminho_pdf_gabarito=gabarito_pdf,
        ano=ano,
        pasta_saida=f"{pasta_saida}/enem_{ano}"
    )
```

```python
# 🔁 LOOP PARA IMPORTAR VÁRIOS ANOS DE UMA VEZ
# Basta colocar os PDFs na pasta 'data/provas_enem/' com o ano no nome (ex: enem_2023_dia2.pdf, enem_2023_gabarito.pdf)
# Dica: O ENEM no formato atual de 45 questões de Matemática (136 a 180) começou em 2009!

for c in range(1998, 2025):
    importar_enem_por_ano(c)
```
